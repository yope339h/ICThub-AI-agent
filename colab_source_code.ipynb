```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# nlp
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize
import string

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

import difflib
import random
import joblib
```

```python
df = pd.read_excel("/content/icthub_dataset.xlsx")
df.head()
```

```python
best_model = joblib.load("/content/chatbot_model.pkl")
label_encoder = joblib.load("/content/label_encoder.pkl")
tfidf_vec = joblib.load("/content/tfidf_vectorizer.pkl")
```

```python
def preprocessing(text):

  text = text.lower()

  text = word_tokenize(text)

  # reomve special characters
  filtered_text = []
  for word in text:
    if word.isalnum():
      filtered_text.append(word)

  text = filtered_text[:]
  filtered_text.clear()

  # remove stopwords and punctuation
  for word in text:
    if word not in stopwords.words('english') and word not in string.punctuation:
      filtered_text.append(word)

  text = filtered_text[:]
  filtered_text.clear()

  # stemming
  ps = PorterStemmer()
  for word in text:
    filtered_text.append(ps.stem(word))

  text = filtered_text[:]
  filtered_text.clear()
  return " ".join(text)
```

```python
user_input = "hello" #input("enter your messgae :")
X_new = preprocessing(user_input)
X_new = tfidf_vec.transform([X_new])
prefiction = best_model.predict(X_new)
predicted_category = label_encoder.inverse_transform(prefiction)[0]
print(predicted_category)
responses = df.groupby("Category")["Chatbot Response"].apply(list).to_dict()
```

```python
def chatbot_response (user_input):
  if not user_input.strip():
    return "your pressed enter witout typing anyhing !"

  user_input = preprocessing(user_input)
  user_input = tfidf_vec.transform([user_input])
  prefiction = best_model.predict(user_input)
  predicted_category = label_encoder.inverse_transform(prefiction)[0]

  responses = df.groupby("Category")["Chatbot Response"].apply(list).to_dict()
  if predicted_category in responses:
    response = random.choice(responses[predicted_category])
    return response.strip()
```

```python
def ask_chatbot(question):
  prompt =f"""
  Here is everything you know. Use ONLY this information to answer.
  If the answer isn't here, say "I don't have that information."
  Answer in plain conversational text only — no markdown, no tables, no asterisks, no HTML tags like <br>.
  answer with the same language

  If the user's message means they want to END the conversation (e.g. "bye", "thanks that's all","thanks", "no more questions"),
  reply normally to say goodbye, then add this exact marker on its own new line at the very end: [END_CHAT]


  {icthub_infromation}
  Question: {question}
  Answer:"""
  resp = requests.post(
    KILOCODE_URL,
    headers={
        "Content-Type": "application/json",
        "Authorization": f"Bearer {KILOCODE_API_KEY}",
    },
    json={
        "model": "kilo-auto/free",
        "messages":[{"role":"user","content":prompt}]
    },
    timeout = 90
  )

  data = resp.json()
  return data['choices'][0]["message"]["content"].strip()
```

```python
import copy
import threading
import uuid
import requests
import gradio as gr

_chat_lock = threading.Lock()

def ask_chatbot(message, knowledge):
    prompt = f"""
Here is everything you know. Use ONLY this information to answer.
If the answer isn't here, say "I don't have that information."
Answer in plain conversational text only — no markdown, no tables, no asterisks, no HTML tags.
Answer in the same language as the question.

{knowledge}

Question: {message}
Answer:"""
    resp = requests.post(
        KILOCODE_URL,
        headers={
            "Content-Type": "application/json",
            "Authorization": f"Bearer {KILOCODE_API_KEY}",
        },
        json={"model": "kilo-auto/free", "messages": [{"role": "user", "content": prompt}]},
        timeout=90,
    )
    data = resp.json()
    if "choices" not in data:
        return f"API error (status {resp.status_code}): {data}"
    return data["choices"][0]["message"]["content"].strip()

def new_id():
    return str(uuid.uuid4())[:8]

def chat_choices(sessions):
    return [(s["title"], sid) for sid, s in sessions.items()]
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    sessions = gr.State({})
    current_id = gr.State(None)
    gr.Markdown(
        """
        # 🤖 ICTHub Assistant
        ### Your virtual guide to ICTHub's digital solutions & internships
        ---
        """
    )

    with gr.Row():
        with gr.Column(scale=1):
            new_btn = gr.Button("+ New chat")
            chat_list = gr.Radio(choices=[], label="Chats")
        with gr.Column(scale=4):
            chatbot = gr.Chatbot(label="ICTHub Assistant")
            msg = gr.Textbox(
                label="Message",
                placeholder="Type your message and press Enter..."
            )

    def start():
        sid = new_id()
        sessions_dict = {sid: {"title": "New chat", "messages": []}}
        return sessions_dict, sid, gr.update(choices=chat_choices(sessions_dict), value=sid), []

    demo.load(start, outputs=[sessions, current_id, chat_list, chatbot])

    def create_new_chat(sessions_dict):
        sessions_dict = copy.deepcopy(sessions_dict)
        sid = new_id()
        sessions_dict[sid] = {"title": "New chat", "messages": []}
        return sessions_dict, sid, gr.update(choices=chat_choices(sessions_dict), value=sid), []

    new_btn.click(create_new_chat, inputs=[sessions], outputs=[sessions, current_id, chat_list, chatbot])

    def switch_chat(sessions_dict, sid):
        if not sid or sid not in sessions_dict:
            return [],sid
        return list(sessions_dict[sid]["messages"]),sid

    chat_list.change(switch_chat, inputs=[sessions, chat_list], outputs=[chatbot,current_id])

    def respond(message, sessions_dict, sid):
        if not message.strip():
            return sessions_dict, gr.update(), "", gr.update()

        with _chat_lock:
            sessions_dict = copy.deepcopy(sessions_dict)

            if not sid or sid not in sessions_dict:
                sid = new_id()
                sessions_dict[sid] = {"title": "New chat", "messages": []}

            sessions_dict[sid]["messages"].append({"role": "user", "content": message})
            reply = ask_chatbot(message, icthub_infromation)
            sessions_dict[sid]["messages"].append({"role": "assistant", "content": reply})

            if sessions_dict[sid]["title"] == "New chat":
                sessions_dict[sid]["title"] = message[:30]

            return (
                sessions_dict,
                list(sessions_dict[sid]["messages"]),
                "",
                gr.update(choices=chat_choices(sessions_dict), value=sid),
            )

    msg.submit(
        respond,
        inputs=[msg, sessions, current_id],
        outputs=[sessions, chatbot, msg, chat_list],
    )

demo.launch(debug=False, share=True)
```
